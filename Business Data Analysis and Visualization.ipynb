```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("2. Finance_Analytics_Dataset.csv")

df
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.isnull().sum()
```

```python
print("Number of duplicates:", df.duplicated().sum())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Category")

plt.title("Number of Transactions in Each Category")
plt.xlabel("Category")
plt.ylabel("Number of Transactions")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="PaymentMode")

plt.title("Number of Transactions by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Number of Transactions")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="TransactionType")

plt.title("Number of Transactions by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
transaction_amount = df.groupby("TransactionType")["Amount"].sum()

plt.figure(figsize=(7, 5))

transaction_amount.plot(kind="bar")

plt.title("Debit vs Credit Amount Comparison")
plt.xlabel("Transaction Type")
plt.ylabel("Total Amount")

plt.xticks(rotation=0)
plt.show()
```

```python
segment_amount = df.groupby("CustomerSegment")["Amount"].sum()

plt.figure(figsize=(8, 5))

segment_amount.plot(kind="bar")

plt.title("Spending Pattern by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Spending")

plt.xticks(rotation=0)
plt.show()
```

```python
payment_counts = df["PaymentMode"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Payment Mode Usage Distribution")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df, x="Category", y="Amount")

plt.title("Boxplot of Transaction Amount by Category")
plt.xlabel("Category")
plt.ylabel("Transaction Amount")

plt.xticks(rotation=45)
plt.show()
```
